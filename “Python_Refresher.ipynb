Question 1

```

# Program to calculate the number of seconds in a given number of days.
# Ask the user to input the number of days
days = int(input("Enter the number of days: "))

# Calculate the number of seconds
# 1 day = 24 hours, 1 hour = 60 minutes, 1 minute = 60 seconds
seconds = days * 24 * 60 * 60

# Display the result
print("{} days have {} seconds.".format(days, seconds))
```

Question 2

```

# Program to calculate the volume of a sphere given its radius.
# Import the math module to use the constant pi
import math

#  input the radius
radius = float(input("Enter the radius of the sphere: "))

# Calculate the volume using the formula: (4/3) * π * r^3
# Use the exponential operator (**) to compute r^3
volume = (4/3) * math.pi * (radius ** 3)

#  result
print(f"The volume of a sphere with radius {radius} is {volume}")
```

Question 4

```
# Program to check if a character is uppercase or lowercase

# Define a function to check the character
def check_case(char):
    # Check if the character is uppercase
    if char.isupper():
        return "Uppercase"
    # Check if the character is lowercase
    elif char.islower():
        return "Lowercase"
    # If it's not a letter
    else:
        return "Not a letter"

# Ask the user for a character
character = input("Enter a single character: ")

# Call the function and store the result
result = check_case(character)

# Show the result
print("The character is:", result)
```

```
# Starting  with initial values
x = 0  # x starts at 0
y = 20 # y starts at 20

# Repeat the steps until y is less than 6
while y >= 6:
    # Subtract 4 from y
    y = y - 4

    # Add 2/y to x
    x = x + (2 / y)

# After the loop ends, display x
print("y is noe less than 6")
print("The value of x is:", x)
```

Question 6

```
# Program to get 5 values from the user, store them in a list, and find the average
7# Creating  an empty list to store the numbers
numbers = []

# Loop 5 times to get user input
for i in range(5):
    # Ask the user for a number
    value = float(input(f"Enter number {i+1}: "))
    # Add the number to the list
    numbers.append(value)

# Calculate the average
total = sum(numbers)      # Add all numbers
average = total / len(numbers)  # Divide by how many numbers

# Display the result
print("The numbers you entered are:", numbers)
print("The average is:", average)
```
